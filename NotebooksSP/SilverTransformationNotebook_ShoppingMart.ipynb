```python
from pyspark.sql.functions import *
from  pyspark.sql.types import *


# read csv files into a dataframe
df_customers = spark.read.format("csv").option("header","true").load("Files/ShoppingMart_BronzeCustomers/ShoppingMart_customers.csv")
df_orders = spark.read.format("csv").option("header","true").load("Files/ShoppingMart_BronzeOrders/ShoppingMart_orders.csv")
df_products = spark.read.format("csv").option("header","true").load("Files/ShoppingMart_BronzeProducts/ShoppingMart_products.csv")

#Clean the data

df_orders = df_orders.dropna(subset = ['OrderID', 'OrderDate', 'CustomerID', 'ProductID', 'TotalAmount'])
df_orders = df_orders.withColumn('OrderDate', to_date(col('OrderDate')))

# Join Orders and Customers table
df_joined_orders = (
    df_orders.join(df_products, on='ProductID', how= 'inner')
    .join(df_customers, on='CustomerID', how = 'inner')
)

display(df_joined_orders)
```

## Writing data to Silvert layer

```python
df_joined_orders.write.mode('overwrite').parquet('Files/ShoppingMart_SilverOrders/ShoppingMart_customers_ordersdata')
```

```python
df_reviews.write.mode('overwrite').parquet('Files/ShoppingMart_SilverReviews/ShoppingMart_reviews')
df_socialmedia.write.mode('overwrite').parquet('Files/ShoppingMart_SilverSocialmedia/ShoppingMart_social_media')
df_weblogs.write.mode('overwrite').parquet('Files/ShoppingMart_SilverWeblogs/ShoppingMart_web_logs')


```
