# <mark> **Gold layer transformations and Aggregations :Shoppingmart data**</mark>

```python
from pyspark.sql.functions import *

orders_df = spark.read.parquet("Files/ShoppingMart_SilverOrders/ShoppingMart_customers_ordersdata")
reviews_df = spark.read.parquet("Files/ShoppingMart_SilverReviews/ShoppingMart_reviews")
socialmedia_df = spark.read.parquet("Files/ShoppingMart_SilverSocialmedia/ShoppingMart_social_media")
weblogs_df = spark.read.parquet("Files/ShoppingMart_SilverWeblogs/ShoppingMart_web_logs")
#display(socialmedia_df)
```

```python
# KPI 1 : Aggregate the web log data to measure user engagement per user on each page and action

weblogs_df = weblogs_df.groupBy('user_id', 'page', 'action').count()
weblogs_df.write.mode('overwrite').parquet('Files/ShoppingMart_GoldWeblogs/ShoppingMart_web_logs')

#display(weblogs_df)
```

```python
# using sql 
weblogs_df.createOrReplaceTempView("weblogs")

results = spark.sql("""SELECT user_id, action, page, COUNT(*) AS count
FROM weblogs
GROUP BY user_id, page, action""")

display(results)
```

```python
# KPI 2 : Aggregates the unstructured social media data to track sentiment trends across differnt platforms

socialmedia_df = socialmedia_df.groupBy('platform', 'sentiment').count()
socialmedia_df.write.mode('overwrite').parquet('Files/ShoppingMart_GoldSocialmedia/ShoppingMart_social_media')
#display(socialmedia_df)
```

```python
# KPI 3 : Aggregate product reviews to calculate the averge rating per product
reviews_df = spark.read.parquet("Files/ShoppingMart_SilverReviews/ShoppingMart_reviews")
reviews_df = reviews_df.groupBy('product_id').agg(round(avg('rating'),2).alias('AvgRating'))
reviews_df.write.mode('overwrite').parquet('Files/ShoppingMart_GoldReviews/ShoppingMart_reviews')
#display(reviews_df)
```
